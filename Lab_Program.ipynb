```python
import pandas as pd
import numpy as np
```

```python
data={
    "Student":["A","B","C","D","E"],
    "Gender":["Male","Female","Male","Female","Male"],
    "StudyHours":[5,6,np.nan,4,7],
    "Attendance":[80,85,75,np.nan,90],
    "Marks":[75,82,68,55,np.nan],
    "Result":["Pass","Pass","Pass","Fail","Pass"]
}
print("\nStep 2: Sample dat created Successfully")
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print(df.isnull().sum())
```

```python
df["StudyHours"]=df["StudyHours"].fillna(
    df["StudyHours"].mean()
)
print(df)
```

```python
df["Marks"]=df["Marks"].fillna(
    df["Marks"].median()
)
print(df)
```

```python
df["Attendance"]=df["Attendance"].ffill()
print(df)
```

```python
df["Gender"]=df["Gender"].map({
    "Male":1,
    "Female":0
})
print(df)
```

```python
df["Result"]=df["Result"].map({
    "Pass":1,
    "Fail":0
})
print(df)
```

```python
print(df.isnull().sum())
```
