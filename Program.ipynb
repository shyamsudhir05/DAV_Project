```python
# DATA COMBINATION FOR BUSINESS UNDERSTANDING (This concept is not
# actual lab program but you have to understand the concept that’s it)
# Now that we have understood and explored the dataset, suppose the business gives us
# additional information stored in different tables. We need to combine that information to
# perform better analysis. This is called Data Combination.
# It covers:
# 1. Concatenation
# 2. Merging
# 3. Joining
# 4. GroupBy
# 5. Pivot Table
```

```python
import pandas as pd
```

```python
# 1. CREATE SALES DATA
df = pd.DataFrame({
    "Order Date": [
        "2024-01-10",
        "2024-01-15",
        "2024-02-05",
        "2024-02-12",
        "2024-03-01",
        "2024-03-15",
        "2024-04-10",
        "2024-04-20",
        "2024-05-05",
        "2024-01-15"
    ],
    "Product Name": [
        "Laptop",
        "Mouse",
        "Printer",
        "Tablet",
        "Laptop",
        "Monitor",
        "Mouse",
        "Printer",
        "Smartwatch",
        "Mouse"
    ],
    "Category": [
        "Electronics",
        "Accessories",
        "Office",
        "Electronics",
        "Electronics",
        "Accessories",
        "Accessories",
        "Office",
        "Electronics",
        "Accessories"
    ],
    "Region": [
        "North",
        "South",
        "East",
        "West",
        "North",
        "South",
        "East",
        "West",
        "North",
        "South"
    ],
    "Quantity": [
        2,
        5,
        3,
        4,
        1,
        2,
        6,
        2,
        3,
        5
    ],
    "Sales": [
        50000,
        2500,
        9000,
        30000,
        55000,
        16000,
        3000,
        6000,
        45000,
        2500
    ],
    "Profit": [
        10000,
        500,
        1500,
        6000,
        11000,
        3000,
        None,
        1000,
        9000,
        500
    ]
})
```

```python
# 2. CONCATENATION

new_sales = pd.DataFrame({
    "Order Date": [
        "2024-06-01",
        "2024-06-05",
        "2024-06-10"
    ],
    "Product Name": [
        "Laptop",
        "Mouse",
        "Printer"
    ],
    "Category": [
        "Electronics",
        "Accessories",
        "Office"
    ],
    "Region": [
        "North",
        "South",
        "East"
    ],
    "Quantity": [
        2,
        4,
        3
    ],
    "Sales": [
        60000,
        2000,
        9000
    ],
    "Profit": [
        12000,
        400,
        1800
    ]
})

print("NEW SALES DATA")

display(new_sales)
combined_df = pd.concat(
    [df, new_sales],
    ignore_index=True
)
print("-"*69)
print("CONCATENATED DATA")
display(combined_df)
```

```python
# 3. MERGING

product_info = pd.DataFrame({
    "Product Name": [
        "Laptop",
        "Mouse",
        "Printer",
        "Tablet",
        "Monitor"
    ],
    "Supplier": [
        "Dell",
        "Logitech",
        "HP",
        "Samsung",
        "LG"
    ],
    "Discount": [
        10,
        5,
        15,
        8,
        12
    ]
})

print("PRODUCT INFORMATION")
display(product_info)
merged_df = pd.merge(
    df,
    product_info,
    on="Product Name",
    how="left"
)
print("-"*88)
print("MERGED DATA")
display(merged_df)
```

```python
# 4. JOINING

region_info = pd.DataFrame(
{
"Region_Manager": [
"Manager A",
"Manager B",
"Manager C",
"Manager D"
]
},

index=[
"North",
"South",
"East",
"West"
]
)

print("REGION INFORMATION")
display(region_info)
df_region = df.set_index("Region")
joined_df = df_region.join(region_info)
print("-"*80)
print("JOINED DATA")
display(joined_df)
```

```python
# 5. GROUPBY

region_sales = df.groupby(
"Region"
)["Sales"].sum()

print("TOTAL SALES BY REGION")
display(region_sales)
```

```python
# 6. PIVOT TABLE

pivot_table = pd.pivot_table(
df,
values="Sales",
index="Category",
columns="Region",
aggfunc="sum",
fill_value=0
)
print("PIVOT TABLE")
display(pivot_table)
```
